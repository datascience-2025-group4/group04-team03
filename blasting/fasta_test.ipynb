```python
def extract_subset_fasta(input_file, output_file, n=100):
    count = 0
    write_seq = False
    with open(input_file) as infile, open(output_file, 'w') as outfile:
        for line in infile:
            if line.startswith(">"):
                count += 1
                if count > n:
                    break
                write_seq = True
            if write_seq:
                outfile.write(line)

extract_subset_fasta("pdb_sequences.fasta", "subset_100.fasta", n=100)
```
