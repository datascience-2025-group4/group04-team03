Multiple Sequence Alignment wird genutzt um all unsere Aminosäure Sequenzen untereinander zu vergleichen. Der Output ist eine Distance-Matrix, welche beschreibt wie sehr zwei Sequenzen 
sich unterscheiden. Ist der Wert z.B. 0.2 sind die Sequenzen zu 80% gleich. Durch diese Methode erhält man numerische Daten, welche man für PCA, k-means clustering etc. verwenden kann.
Das Ziel ist, cluster auf Aminosäure Ebene zu finden und zu überprüfen ob sie mit den Genfamilien übereinstimmen.

```python
import os
import platform
import urllib.request
import zipfile
import tarfile

def download_clustalo(dest_dir="clustalo_bin"):
    os.makedirs(dest_dir, exist_ok=True)
    system = platform.system()

    if system == "Windows":
        url = "https://www.ebi.ac.uk/Tools/services/web/toolform.ebi?tool=clustalo&sequence=example"
        print("Bitte Clustal Omega manuell für Windows herunterladen:")
        print("🔗 https://www.ebi.ac.uk/Tools/msa/clustalo/")
        print("Entpacke es und füge den Pfad zur .exe in die Umgebungsvariablen ein.")
        return

    elif system == "Linux":
        url = "http://www.clustal.org/omega/clustal-omega-1.2.4.tar.gz"
    elif system == "Darwin":  # macOS
        url = "http://www.clustal.org/omega/clustal-omega-1.2.4-macosx.zip"
    else:
        print("Unbekanntes Betriebssystem.")
        return

    filename = url.split("/")[-1]
    filepath = os.path.join(dest_dir, filename)

    print(f"⬇️  Lade Clustal Omega von {url}")
    urllib.request.urlretrieve(url, filepath)

    print("📦 Entpacke...")
    if filename.endswith(".zip"):
        with zipfile.ZipFile(filepath, 'r') as zip_ref:
            zip_ref.extractall(dest_dir)
    elif filename.endswith(".tar.gz"):
        with tarfile.open(filepath, "r:gz") as tar_ref:
            tar_ref.extractall(dest_dir)

    print(f"✅ Fertig! Clustal Omega befindet sich in: {dest_dir}")

download_clustalo()
```
