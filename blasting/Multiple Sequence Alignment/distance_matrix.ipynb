In this code, the results from the MSA will be turned into a distance matrix to obtain numerical values and then perform PCA and K-means clustering.

```python
from Bio import AlignIO
from Bio.Align import MultipleSeqAlignment
import pandas as pd
import numpy as np
```

```python
alignment = AlignIO.read("aligned_annotated_antibody_sequences.aln", "clustal")
 #Saving file under "alignment

def seq_distance(seq1, seq2):
    matches = sum((a == b) for a, b in zip(seq1, seq2) if a != '-' and b != '-')
    valid_positions = sum((a != '-' and b != '-') for a, b in zip(seq1, seq2))
    if valid_positions == 0:
        return 1.0  # Maximum distance if no overlap
    return 1 - (matches / valid_positions)

labels = [record.id for record in alignment]
n = len(alignment)
matrix = np.zeros((n, n))

for i in range(n):
    for j in range(i + 1, n):
        dist = seq_distance(alignment[i].seq, alignment[j].seq)
        matrix[i, j] = matrix[j, i] = dist

df = pd.DataFrame(matrix, index=labels, columns=labels)
df.to_csv("antibodies_distance_matrix.csv")

```
