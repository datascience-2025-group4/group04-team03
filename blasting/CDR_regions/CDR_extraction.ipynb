```python
import pandas as pd
from anarci import anarci

# CSV einlesen
df = pd.read_csv("../../documentation/pdb_sequences_detailed.csv")

def get_cdr_regions(seq):
    try:
        results, _ = anarci([("seq", seq)], scheme="imgt", assign_germline=False)
    except Exception as e:
        print(f"Fehler bei Sequenz: {seq[:10]}... - {e}")
        return None

    # results ist Liste mit Einträgen, wir nehmen den ersten
    # Er hat die Form: (annotation, alignment, germline)
    if not results or results[0][0] is None:
        return None

    annotation = results[0][0]  # annotation ist Liste von (residue_number, position_label, aa)

    cdr1 = []
    cdr2 = []
    cdr3 = []

    for resnum, pos_label, aa in annotation:
        if pos_label is None:
            continue
        # Extract numeric part from pos_label (e.g. '27A' -> 27)
        try:
            number = int(''.join(filter(str.isdigit, pos_label)))
        except ValueError:
            continue

        if 27 <= number <= 38:
            cdr1.append(aa)
        elif 56 <= number <= 65:
            cdr2.append(aa)
        elif 105 <= number <= 117:
            cdr3.append(aa)

    return {
        "CDR1": "".join(cdr1),
        "CDR2": "".join(cdr2),
        "CDR3": "".join(cdr3)
    }

# CDRs extrahieren
cdr_data = df['sequence'].apply(get_cdr_regions)

df['CDR1'] = cdr_data.apply(lambda x: x['CDR1'] if x else None)
df['CDR2'] = cdr_data.apply(lambda x: x['CDR2'] if x else None)
df['CDR3'] = cdr_data.apply(lambda x: x['CDR3'] if x else None)

df.to_csv("pdb_sequences_with_CDRs.csv", index=False)

print("Fertig: CDR-Regionen extrahiert und gespeichert.")
```

```python
from anarci import anarci

seq = "QVQLVQSGAEVKKPGSSVKVSCKTSGDTFSTYAISWVRQAPGQGLEWMGGIIPIFGKAHYAQKFQGRVTITADESTSTAYMELSSLRSEDTAVYFCARKFHFVSGSPFGMDVWGQGTTVTVSSASTKGPSVFPLAPSSKSTSGGTAALGCLVKDYFPEPVTVSWNSGALTSGVHTFPAVLQSSGLYSLSSVVTVPSSSLGTQTYICNVNHKPSNTKVDKKVEPKSCDKTHHHHHH"

results = anarci([("seq", seq)], scheme="imgt", assign_germline=True)

residues = results[0][0][0][0]

cdr_regions = {"CDR1": [], "CDR2": [], "CDR3": []}

for (res_num, ins_code), aa in residues:
    if aa == '-':
        continue  # Lücken überspringen

    res_number = str(res_num) + ins_code  # z. B. '27', '27A', etc.

    # Einfache Bereichsabfragen (auch mit Insertionscode)
    if 27 <= res_num <= 38:
        cdr_regions["CDR1"].append(aa)
    elif 56 <= res_num <= 65:
        cdr_regions["CDR2"].append(aa)
    elif 105 <= res_num <= 117:
        cdr_regions["CDR3"].append(aa)

# Ausgabe
for cdr, seq in cdr_regions.items():
    print(f"{cdr}: {''.join(seq)}")
```

```python
# Residues mit Positionen und AAs (ganze Liste)
residues = results[0][0][0]

# Dictionary mit Positionslabels
pos_labels = results[0][0][1]

cdr_regions = {"CDR1": [], "CDR2": [], "CDR3": []}

for (res_num, ins_code), aa in residues:
    if aa == '-':
        continue

    # Hole Label aus dem Dictionary
    pos_label = pos_labels.get((res_num, ins_code), "")

    if pos_label.startswith("CDR1"):
        cdr_regions["CDR1"].append(aa)
    elif pos_label.startswith("CDR2"):
        cdr_regions["CDR2"].append(aa)
    elif pos_label.startswith("CDR3"):
        cdr_regions["CDR3"].append(aa)

for cdr, seq in cdr_regions.items():
    print(f"{cdr}: {''.join(seq)}")
```
