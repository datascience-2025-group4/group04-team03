```python
from anarci import anarci

seq = "QVQLVQSGAEVKKPGSSVKVSCKTSGDTFSTYAISWVRQAPGQGLEWMGGIIPIFGKAHYAQKFQGRVTITADESTSTAYMELSSLRSEDTAVYFCARKFHFVSGSPFGMDVWGQGTTVTVSSASTKGPSVFPLAPSSKSTSGGTAALGCLVKDYFPEPVTVSWNSGALTSGVHTFPAVLQSSGLYSLSSVVTVPSSSLGTQTYICNVNHKPSNTKVDKKVEPKSCDKTHHHHHH"

results = anarci([("seq", seq)], scheme="imgt", assign_germline=True)

residues = results[0][0][0][0]

cdr_regions = {"CDR1": [], "CDR2": [], "CDR3": []}

for (res_num, ins_code), aa in residues:
    if aa == '-':
        continue  # Lücken überspringen

    res_number = str(res_num) + ins_code  # z. B. '27', '27A', etc.

    # Einfache Bereichsabfragen (auch mit Insertionscode)
    if 27 <= res_num <= 38:
        cdr_regions["CDR1"].append(aa)
    elif 56 <= res_num <= 65:
        cdr_regions["CDR2"].append(aa)
    elif 105 <= res_num <= 117:
        cdr_regions["CDR3"].append(aa)

# Ausgabe
for cdr, seq in cdr_regions.items():
    print(f"{cdr}: {''.join(seq)}")
```

```python
import pandas as pd

# ---------- Hilfsfunktion zum sicheren Parsen der Entity-IDs ----------
def parse_entity_ids(cell):
    if pd.isna(cell):
        return []
    if isinstance(cell, str):
        return [int(e.strip()) for e in cell.replace(',', '|').split('|') if e.strip().isdigit()]
    elif isinstance(cell, (int, float)):
        return [int(cell)]
    return []

# ---------- CSV-Dateien laden ----------
mapping_df = pd.read_csv("../../documentation/clean_entity_mapping.csv")
seq_df = pd.read_csv("../../documentation/pdb_sequences_detailed.csv")

# ---------- Ergebnisliste vorbereiten ----------
results = []

# ---------- Alle PDBs im Mapping durchgehen ----------
for idx, row in mapping_df.iterrows():
    pdb_id = row["pdb"]
    h_ids = parse_entity_ids(row.get("H_entity_id", ""))
    l_ids = parse_entity_ids(row.get("L_entity_id", ""))

    # Kombiniere alle gewünschten Entity-IDs
    relevant_ids = set(h_ids + l_ids)
    print(f"🔍 Verarbeite {pdb_id} mit Entity-IDs: {relevant_ids}")

    # Finde passende Sequenzdaten
    subset = seq_df[(seq_df["pdb_id"] == pdb_id) & (seq_df["entity_id"].isin(relevant_ids))]

    if subset.empty:
        print(f"⚠️ Keine passenden Einträge gefunden für {pdb_id}.")
    else:
        results.append(subset)

# ---------- Ergebnisse zusammenfügen ----------
if results:
    final_df = pd.concat(results, ignore_index=True)
    final_df.to_csv("antibody_entities_only.csv", index=False)
    print(f"\n✅ Gespeichert: {len(final_df)} Einträge in 'antibody_entities_only.csv'")
else:
    print("❌ Keine passenden Einträge gefunden.")
```

```python
import pandas as pd
from anarci import anarci

# CSV mit Sequenzen laden
df = pd.read_csv("antibody_entities_only.csv")

def extract_cdrs_from_sequence(seq_id, sequence):
    """Extract CDRs from a single sequence using ANARCI, handling both heavy and light chains"""
    try:
        # Call ANARCI
        results = anarci([("seq", sequence)], 
                        scheme="imgt", 
                        allowed_species=["human", "mouse", "rat", "rabbit"])
        
        if not results or not results[0] or not results[0][0]:
            return "", "", "", "", "No ANARCI results"
        
        # Get the numbering data and chain information
        chain_data = results[0][0][0]  # (numbering_list, start_idx, end_idx)
        numbering_list = chain_data[0]  # The actual list of numbered residues
        
        # Get alignment details to determine chain type
        alignment_details = results[1] if len(results) > 1 else None
        
        if not numbering_list:
            return "", "", "", "", "No numbering data"
        
        # Try to determine chain type from ANARCI results
        # ANARCI sometimes includes chain type information in the results
        chain_type = "Unknown"
        
        # Method 1: Check if we can get chain type from alignment details
        if alignment_details and len(alignment_details) > 0:
            try:
                # alignment_details structure can vary, let's be safe
                first_alignment = alignment_details[0]
                if hasattr(first_alignment, 'get') or isinstance(first_alignment, dict):
                    chain_type = first_alignment.get('chain_type', 'Unknown')
                elif isinstance(first_alignment, (list, tuple)) and len(first_alignment) > 0:
                    # Sometimes chain type info is in different positions
                    for item in first_alignment:
                        if isinstance(item, str) and item in ['H', 'L', 'K']:
                            chain_type = item
                            break
            except:
                pass
        
        # Method 2: Infer chain type from sequence characteristics and position ranges
        if chain_type == "Unknown":
            # Get the position range to help identify chain type
            positions = [pos for (pos, insertion), residue in numbering_list if residue != "-"]
            if positions:
                max_pos = max(positions)
                # Heavy chains typically go up to ~127-130, light chains usually ~107-109
                if max_pos > 120:
                    chain_type = "H"  # Heavy chain
                else:
                    chain_type = "L"  # Light chain (could be kappa or lambda)
        
        # Define CDR positions based on IMGT numbering scheme
        if chain_type == "H":  # Heavy chain
            cdr_ranges = {
                "CDR1": set(range(27, 39)),   # 27-38
                "CDR2": set(range(56, 66)),   # 56-65  
                "CDR3": set(range(105, 118)), # 105-117
            }
        else:  # Light chain (both kappa and lambda have same CDR positions in IMGT)
            cdr_ranges = {
                "CDR1": set(range(27, 39)),   # 27-38
                "CDR2": set(range(56, 66)),   # 56-65 (though often shorter)
                "CDR3": set(range(105, 118)), # 105-117 (though often shorter)
            }
            # Note: Light chain CDR2 is often shorter (56-65 range but may not use all positions)
            # and CDR3 is typically shorter than heavy chain CDR3
        
        # Initialize CDR residue collections
        cdr1_residues = []
        cdr2_residues = []
        cdr3_residues = []
        
        # Process each numbered residue
        for position_info, residue in numbering_list:
            # position_info is (position, insertion)
            position, insertion = position_info
            
            # Skip gaps
            if residue == "-":
                continue
            
            # Assign to appropriate CDR based on position
            if position in cdr_ranges["CDR1"]:
                cdr1_residues.append((position, residue))
            elif position in cdr_ranges["CDR2"]:
                cdr2_residues.append((position, residue))
            elif position in cdr_ranges["CDR3"]:
                cdr3_residues.append((position, residue))
        
        # Sort by position and extract sequences
        cdr1_residues.sort(key=lambda x: x[0])
        cdr2_residues.sort(key=lambda x: x[0])
        cdr3_residues.sort(key=lambda x: x[0])
        
        cdr1_seq = "".join([res for pos, res in cdr1_residues])
        cdr2_seq = "".join([res for pos, res in cdr2_residues])
        cdr3_seq = "".join([res for pos, res in cdr3_residues])
        
        # Create status message
        status = f"Success: {chain_type} chain, CDR1({len(cdr1_seq)}) CDR2({len(cdr2_seq)}) CDR3({len(cdr3_seq)})"
        
        return cdr1_seq, cdr2_seq, cdr3_seq, chain_type, status
        
    except Exception as e:
        import traceback
        error_details = traceback.format_exc()
        return "", "", "", "", f"Error: {str(e)}"

# Process all sequences
cdr1_list = []
cdr2_list = []
cdr3_list = []
chain_type_list = []
status_list = []

print("Processing sequences...")
for idx, row in df.iterrows():
    seq_id = row.get("pdb_id", f"seq_{idx}")
    sequence = row["sequence"]
    
    cdr1, cdr2, cdr3, chain_type, status = extract_cdrs_from_sequence(seq_id, sequence)
    
    cdr1_list.append(cdr1)
    cdr2_list.append(cdr2)
    cdr3_list.append(cdr3)
    chain_type_list.append(chain_type)
    status_list.append(status)
    
    # Print progress
    if "Success" in status and (cdr1 or cdr2 or cdr3):
        print(f"✅ {seq_id}: {chain_type} chain - CDR1='{cdr1}' CDR2='{cdr2}' CDR3='{cdr3}'")
    elif "Success" in status:
        print(f"⚠️  {seq_id}: {chain_type} chain - Success but no CDRs found")
    else:
        print(f"❌ {seq_id}: {status}")
    
    # Progress indicator
    if (idx + 1) % 100 == 0:
        print(f"Progress: {idx + 1}/{len(df)} sequences processed")

# Add results to DataFrame
df["CDR1"] = cdr1_list
df["CDR2"] = cdr2_list
df["CDR3"] = cdr3_list
df["chain_type"] = chain_type_list
df["extraction_status"] = status_list

# Save results
output_file = "pdb_sequences_with_cdrs_.csv"
df.to_csv(output_file, index=False)

# Summary statistics
print(f"\n{'='*50}")
print("EXTRACTION SUMMARY")
print(f"{'='*50}")
print(f"Total sequences processed: {len(df)}")
print(f"Successful extractions: {sum(1 for s in status_list if 'Success' in s)}")

# Chain type distribution
from collections import Counter
chain_counts = Counter(chain_type_list)
print(f"\nChain type distribution:")
for chain_type, count in chain_counts.items():
    print(f"  {chain_type} chains: {count}")

# CDR extraction success by chain type
print(f"\nCDR extraction success:")
heavy_chains = [i for i, ct in enumerate(chain_type_list) if ct == 'H']
light_chains = [i for i, ct in enumerate(chain_type_list) if ct == 'L']

if heavy_chains:
    h_cdr1 = sum(1 for i in heavy_chains if cdr1_list[i])
    h_cdr2 = sum(1 for i in heavy_chains if cdr2_list[i])
    h_cdr3 = sum(1 for i in heavy_chains if cdr3_list[i])
    print(f"  Heavy chains ({len(heavy_chains)} total): CDR1={h_cdr1}, CDR2={h_cdr2}, CDR3={h_cdr3}")

if light_chains:
    l_cdr1 = sum(1 for i in light_chains if cdr1_list[i])
    l_cdr2 = sum(1 for i in light_chains if cdr2_list[i])
    l_cdr3 = sum(1 for i in light_chains if cdr3_list[i])
    print(f"  Light chains ({len(light_chains)} total): CDR1={l_cdr1}, CDR2={l_cdr2}, CDR3={l_cdr3}")

# Show some examples by chain type
print(f"\nExample extractions:")
heavy_examples = 0
light_examples = 0

for i, (seq_id, cdr1, cdr2, cdr3, chain_type, status) in enumerate(zip(
    df["pdb_id"], cdr1_list, cdr2_list, cdr3_list, chain_type_list, status_list)):
    
    if "Success" in status and (cdr1 or cdr2 or cdr3):
        if chain_type == "H" and heavy_examples < 3:
            print(f"Heavy chain {seq_id}: CDR1='{cdr1}' CDR2='{cdr2}' CDR3='{cdr3}'")
            heavy_examples += 1
        elif chain_type == "L" and light_examples < 3:
            print(f"Light chain {seq_id}: CDR1='{cdr1}' CDR2='{cdr2}' CDR3='{cdr3}'")
            light_examples += 1
    
    if heavy_examples >= 3 and light_examples >= 3:
        break

print(f"\nResults saved to: {output_file}")
```
