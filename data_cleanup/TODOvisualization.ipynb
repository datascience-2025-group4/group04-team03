Barplots and heatmaps

```python
# For creating statistical visualizations
import seaborn as sns
# For data manipulation and analysis
import pandas as pd
# For numerical operations
import numpy as np
```

```python
# Clean white style
sns.set_style("white")

df = pd.read_csv("../data/ab_ag.tsv", sep="\t")

sns.barplot(
    data = df,      # Your DataFrame
    x= df["light_ctype"],         # Category variable
    y= df["r_factor"],         # Numeric variable
    estimator='mean',  # Statistical function to estimate
    errorbar=('ci', 95),  # Error bars type and level
    orient="v"     # "v" for vertical, "h" for horizontal
)
```
